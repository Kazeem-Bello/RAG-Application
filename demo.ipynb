```python
import json

with open (f"{file_path}/files/grandstay_hotel_knowledge_base.json", "r") as file:
    data = json.load(file)
```

```python
for i in data['hotel_knowledge_base']:
    print(i)
```

```python
documents = []
for knowledge_base in data['hotel_knowledge_base']:
    doc = Document(
        page_content=(
            f"category: {knowledge_base["category"]}\n"
            f"title: {knowledge_base["title"]}\n"
            f"title: {knowledge_base["content"]}"
        ),
        metadata={
            "source": f"{file_path}/files/grandstay_hotel_knowledge_base.json"
        }
    )
    documents.append(doc)
    
```

```python
file_paths = [
    "files/City_Insider_Guide.pdf",
    "files/GrandStay_Amenities_Guide.pdf",
    "files/GrandStay_Policies_Handbook.pdf",
    "files/GrandStay_Staff_SOP.pdf",
    "files/grandstay_hotel_knowledge_base.json"
]

absolute_path = Path.cwd()
```

```python
documents = []
for file_path in file_paths:
    ingestion = Ingestion(file_path=f"{absolute_path}/{file_path}")
    if Path(file_path).suffix == ".pdf":
        pages = ingestion.load_pdf_file()
        chunks = ingestion.chunking(pages)
        documents.extend(chunks)
    if Path(file_path).suffix == ".json":
        chunks = ingestion.load_json_file()
        documents.extend(chunks)
```

```python
vector_store = ingestion.save_to_vector_store(documents)
```
